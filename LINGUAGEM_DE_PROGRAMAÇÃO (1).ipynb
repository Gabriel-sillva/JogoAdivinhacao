# Operadores Condicionais e Lógicos em Python

Os operadores condicionais e lógicos são fundamentais para controlar o fluxo de execução do seu código através de estruturas como `if`, `elif` e `else`.

---

## 1. Operadores de Comparação (Relacionais)

Estes operadores comparam dois valores e retornam um valor booleano (`True` ou `False`).

| Operador | Sinal | Notação Matemática | Descrição | Exemplo (`a = 5`, `b = 10`) | Resultado |
| :--- | :---: | :---: | :--- | :--- | :---: |
| **Menor que** | `<` | $a < b$ | Estritamente menor que | `a < b` | `True` |
| **Menor ou igual** | `<=` | $a \le b$ | Menor ou igual a | `a <= b` | `True` |
| **Maior que** | `>` | $a > b$ | Estritamente maior que | `a > b` | `False` |
| **Maior ou igual** | `>=` | $a \ge b$ | Maior ou igual a | `a >= b` | `False` |
| **Igual a** | `==` | $a = b$ | Igualdade de valor | `a == b` | `False` |
| **Diferente de** | `!=` | $a \neq b$ | Diferença de valor | `a != b` | `True` |

---

## 2. Operadores de Identidade

Utilizados para verificar se duas variáveis apontam para o mesmo objeto na memória, e não apenas se possuem o mesmo valor.

* `is`: Retorna `True` se ambas as variáveis forem o **mesmo objeto**.
* `is not`: Retorna `True` se as variáveis **não forem o mesmo objeto** (Negação da identidade).

> **Nota:** Use `==` para comparar **valores** e `is` para comparar **identidade de memória** (muito usado com `None`).

---

## 3. Operadores Lógicos

Permitem combinar múltiplas expressões condicionais.

| Operador Python | Significado | Descrição Lógica | Exemplo |
| :---: | :---: | :--- | :--- |
| `and` | **E** | Retorna `True` apenas se **todas** as condições forem verdadeiras | `(5 > 3) and (10 > 5)` $\rightarrow$ `True` |
| `or` | **OU** | Retorna `True` se pelo menos **uma** das condições for verdadeira | `(5 > 3) or (10 < 5)` $\rightarrow$ `True` |
| `not` | **NÃO** | Inverte o resultado lógico (Negação) | `not(5 > 3)` $\rightarrow$ `False` |

---

#ESTRUTURAS DE REPETIÇÃO EM PYTHON (FOR, WHILE, RANGE, BREAK, CONTINUE)
--------------------------------------------------------------------------------

#1. FUNÇÃO RANGE()

    A função range() gera uma sequência de números. É amplamente utilizada com o
    laço FOR para repetir um bloco de código um número determinado de vezes.

    Sintaxe: range(início, fim, passo)
      - início: valor inicial (inclusivo). Padrão é 0.
      - fim: valor final (exclusivo).
      - passo: incremento/decremento entre os números. Padrão é 1.

--------------------------------------------------------------------------------

# Exemplo 1.1: Gerando números de 0 a 4
    print("--- range(5) ---")
    for i in range(5):
        print(i)

--------------------------------------------------------------------------------

# Exemplo 1.2: Especificando início e fim (5 até 9)
    print("\n--- range(5, 10) ---")
    for i in range(5, 10):
        print(i)

-------------------------------------------------------------------------------

# Exemplo 1.3: Especificando o passo (de 0 a 10 de 2 em 2)
    print("\n--- range(0, 11, 2) ---")
    for i in range(0, 11, 2):
        print(i)

--------------------------------------------------------------------------------

# Exemplo 1.4: Contagem regressiva (de 10 até 1)
    print("\n--- range(10, 0, -1) ---")
    for i in range(10, 0, -1):
        print(i)

#5. INSTRUÇÃO CONTINUE
-------------------------------------------------------------------------------
    A instrução CONTINUE pula o restante do código dentro do laço APENAS para a

    # Exemplo 5.1: Imprimindo apenas números ímpares (pula os pares)

    for i in range(1, 10):

    if i % 2 == 0:  # Se for par
        continue    # Pula o resto da iteração
    print(f"Número ímpar: {i}")

````python
# Exemplo simples de loop WHILE que verifica se um número inserido pelo usuário
# é par ou ímpar e encerra o programa quando o número zero for inserido

# Primeira leitura antes de entrar no loop
numero = int(input("Digite um número (ou 0 para sair): "))

while numero != 0:
    # Verifica se o número é par
    if numero % 2 == 0:
        print(f"O número {numero} é PAR.")
    else:
        print(f"O número {numero} é ÍMPAR.")

    # Pede o próximo número dentro do loop
    numero = int(input("Digite outro número (ou 0 para sair): "))

print("Programa encerrado!")
````

````python
# Método 1 – Repetição por quantidade

# No primeiro método, você pode passar um único argumento para range(),
# que representa a quantidade de vezes em que o loop deve ser repetido.
# Por exemplo, range(5) cria uma sequência que se inicia em 0 e vai até 4, realizando 5 iterações.

for x in range(5):
    # 'x' assume os valores de 0 até 4 a cada iteração
    print(x)
````

````python
# Método 2 – Limites inicial e superior

# No segundo método, você pode fornecer dois argumentos para range().
# O primeiro argumento representa o início das repetições e o segundo,
# o limite superior (não incluso) do valor da variável de controle. Por exemplo,
# range(2, 7) cria uma sequência que se inicia em 2 e vai até 6, realizando 5 iterações.

for y in range(2, 7):

    print(y) # linha Recuada para Fazer parte do bloco do loop
````
